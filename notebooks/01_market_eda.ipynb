# Used Car Market EDA
This notebook performs portfolio-ready exploratory analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('../data/used_cars_demo.csv')
df.head()
```

```python
df.info()
df.describe()
```

```python
df.groupby('brand')['price_lakh'].agg(['count','mean','median']).sort_values('mean',ascending=False)
```

```python
sns.scatterplot(data=df,x='kilometers',y='price_lakh',hue='fuel',alpha=.45)
plt.title('Price vs Mileage'); plt.show()
```

```python
sns.boxplot(data=df,x='fuel',y='price_lakh')
plt.title('Price Distribution by Fuel'); plt.show()
```
